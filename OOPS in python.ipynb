# Define of OOPS:-

OOPs stands for Object-Oriented Programming System. It is a programming approach based on classes and objects. A class is a blueprint used to create objects, while an object represents a particular instance of a class. OOPs helps organize code into reusable and manageable parts. Python supports four main principles of OOPs: encapsulation, inheritance, polymorphism, and abstraction.

1. Class and Object:-
A class define attribute and methods .An object is created from that class and can use its properties and bheavior.

2.Encapsulation:- encapsulation combine data and methods inside a class and helps control access to the data.

3.Inhertance :-
Inheritance allow a child to reuse and extend the attributes and methods of a parents class

4.PLoymorphism :- Polymorphism means one interface or methods name can have different bheviors for different objects.

5.Abstraction:- Abstraction hides implement details and exposes only the essestional feature neended by the user.

# Advantage of OOPs:-

- Code reusability: Inheritance helps reuse existing code.

- Data organization: Classes keep related data and methods together.

- Maintainability: Code can be divided into manageable classes.

- Flexibility: Polymorphism supports different implementations through a common interface.

- Data control: Encapsulation helps manage access to object data.

# Disadvantage of OOPs:-

- Complexity: OOPs can be difficult for beginners to understand.

- More code: Small programs may require extra classes and methods.

- Memory overhead: Objects and their attributes consume memory.

- Design challenges: Poor class design can make programs difficult to maintain.

- Learning time: Understanding all four principles takes practice.

```python
from abc import ABC,abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Dog says woof")

class Cat(Animal):
    def sound(self):
        print(" Cat says meow")

d=Dog()
d.sound()

c=Cat()
c.sound()
```

```python
from abc import ABC,abstractmethod
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass
class Car(Vehicle):
    def start(self):
        print(" car start with key")

class Bike(Vehicle):
    def start(self):
        print("bike start with button ")

c=Car()
c.start()

b=Bike()
b.start()   
```
