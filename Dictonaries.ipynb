# Dictonary :-

A dictionary in Python is a data type used to store data in key-value pairs. Each key is used to access its corresponding value. Dictionaries are written using curly brackets {}. Keys must be unique, but values can be duplicated. Dictionaries are mutable, which means we can add, update, and delete items after creating them.

# Advantage:-

- It stores related data in an organized way.
- It allows quick access to values using keys.
- We can easily add, update, and delete items.
- It can store different types of data together.

# Disadvantage:-

- Dictionaries use more memory than some simpler data structures.

- Duplicate keys are not allowed; a repeated key overwrites its previous value.

- Accessing a missing key directly can raise a KeyError.

- Dictionaries are not accessed by numerical position like lists.

```python
student = {"name": "Raj", "age": 21}
print(student["name"])
```

```python
student = {"name": "Raj", "age": 21}
student["course"] = "BCA"
print(student)
```

```python
student = {"name": "Raj", "age": 21}
student["age"] = 22
print(student)
```

```python
student = {"name": "Raj", "age": 21}
del student["age"]
print(student)
```

```python
student = {"name": "Raj", "age": 21}

for key, value in student.items():
    print(key, value)
```

```python
student = {"name": "Raj", "age": 21}

print(student.get("name"))
print(student.get("marks", 0))
```
