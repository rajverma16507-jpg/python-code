# File handling:-

File handling in Python is used to create, open, read, write, and modify files. It allows us to store data permanently on a computer, even after the program has stopped running. Python provides the built-in open() function to work with files. We can handle text files, such as .txt, and other file formats. File handling is useful for storing student records, user information, logs, and reports,user information ,logs,and reports.

# Advantage:-

- ermanent storage: Data remains saved after the program ends.

- Easy access: We can read stored data whenever required.

- Data sharing: Files can transfer information between programs.

- Large data storage: Files can store more data than we would conveniently keep in program variables.

- Data management: We can add, update, and delete file contents using appropriate operations.

# Disadavantage:-

- File errors: A program may fail if a file does not exist or cannot be accessed.

- Security risks: Sensitive information can be exposed if files are not protected.

- Data loss: Incorrect write operations can overwrite existing data.

- Performance: Reading and writing very large files can take time.

- Limited structure: Plain text files do not provide the structured querying and relationships available in databases.

# File opening Models:-

- r = read an existing file.
- w = write to a file,overwrites existing conent or creates a file.
- a = append  data at the end of a file.
- x = create a new file,raise an error if it already exists.
- b = binary mode,used with models such as rb and wb.

# Examples:-

```python
file = open("student.txt", "w")
file.write("My name is Raj")
file.close()
```

```python
file = open("student.txt", "r")
data = file.read()
print(data)
file.close()
```

```python
file = open("student.txt", "a")
file.write("\nI am a BCA student")
file.close()
```

```python
with open("student.txt", "r") as file:
    for line in file:
        print(line.strip())
```

```python
with open("student.txt", "r") as file:
    data = file.read()
    words = data.split()
    print("Total words:", len(words))
```

```python
with open("student.txt", "r") as file:
    print(file.tell())
    file.read(4)
    print(file.tell())
```

```python
with open("student.txt", "r") as file:
    file.seek(0)
    print(file.read(4))
```

```python
try:
    with open("unknown.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File does not exist")
```
