# Define of loops :-

A loop in Python is used to execute a block of code repeatedly. It helps us perform the same task multiple times without writing the same code again and again. Python provides two main types of loops: for loop and while loop. Loops can work with sequences, ranges, and conditions. They make programs shorter, easier to understand, and more efficient.

# Type of loops in python:-

# 1. for loops :-

A for loop is used to iterate over a sequesnce ,such as a list ,string,tuples or range.

# 2.while loops :-

A while loops replace a block  of code as long as its condition remains true.

# Advantage of loops:-

- Code reusability: Repeats instructions without duplicating code.

- Less code: Reduces repetitive statements.

- Automation: Performs repetitive calculations automatically.

- Easy data processing: Iterates over lists, strings, tuples, and dictionaries.

- Improved efficiency: Makes repeated operations easier to implement and maintain.

# Disadvantage of loops :-

- Infinite loops: A wrong condition or missing counter update can make a loop run indefinitely.

- Debugging difficulty: Complex nested loops can be difficult to understand and troubleshoot.

- Performance issues: Very large or inefficient loops can take a long time to execute.

- Logic errors: Incorrect conditions or counter updates can produce wrong results.

- Excessive nesting: Too many nested loops can make code harder to read and maintain.

```python
for i in range(1, 6):
    print(i)
```

```python
fruits = ["Apple", "Mango", "Banana"]

for fruit in fruits:
    print(fruit)
```

```python
total = 0

for i in range(1, 6):
    total += i

print(total)
```

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
i = 2

while i <= 10:
    print(i)
    i += 2
```

```python
n = 5
fact = 1

while n > 0:
    fact *= n
    n -= 1

print(fact)
```

```python
for i in range(1, 6):
    if i == 3:
        break
    print(i)
```

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
for i in range(1, 4):
    print(i)
else:
    print("Loop completed")
```

```python
fruits = ["Apple", "Mango", "Banana"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```
