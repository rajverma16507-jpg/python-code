# Break Statement :-

The break statement is used to stop a loop immediately.when a specific condition is statisfied,the loop teminated .it can be used in both for and while loops. after executing break, the program continue with the statement  after the loops .it helps avoid unneccesary iterations and saves executions time.

# Advantage:-

- Stop the loop immediately when required .
- Save time by avoiding unnecessary iterations.
- Make it easy to search for a particular value .

# Disadvantage:-

- it may stop the loop before all item are processed.
- incoorect use can some required operations to be skipped.    

```python
for i in range(1,11):
    if i==5:
        break
    print(i)
```

```python
i=1
while i<=10:
    if i==4:
        break
    print(i)
    i+=1
```

```python
num=[10,20,30,40]
for n in num:
    if n==30:
        print("found")
        break
```

```python
for n in range(1,20):
    if n%2==0:
        break
    print(n)
```

# Continue Statement:-

The continue statement is used to skip the current itreations of a loop. when a specific conditions is satisfied ,it takes the reamining code dor that iterations . it does not stop the entire loop. After continue ,the loop moves to the next iterations .it can used in both for and while loops.

    

# Advantage:-

- it skip unwanted values without stopping the loop.
- it helps filter data according to a conditions.
- it allows the loop to process the remaining values.    

# Disadvantage:-

- It may skip important code if used incorrectly.
- In a while loop,it can an infinite loop if the counter is not updated before  continue.    

```python
for  i in range(1,11):
    if i==6:
        continue 
    print(i)
```

```python
for i in range(1,11):
    if i%2==0:
        continue 
    print(i)
```

```python
numbers = [1, 0, 2, 0, 3]

for n in numbers:
    if n == 0:
        continue
    print(n)
```

```python
for ch in "RAAJ":
    if ch == "A":
        continue
    print(ch)
```

```python
for i in range(1, 11):
    if i % 3 == 0:
        continue
    print(i)
```
