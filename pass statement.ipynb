# Define of Pass:-

The pass statement in Python is a null statement that performs no operation. It is used as a placeholder when a statement is required syntactically but no action needs to be performed yet. It does not terminate a loop or skip an iteration. The program continues executing the next statement normally. It is commonly used in loops, functions, classes, and conditional statements during development.

# Advantage:-

- It allows us to create an empty function or class temporarily.

- It prevents syntax errors where Python requires a statement.

- It helps developers plan code structure before implementing functionality.

# disadvantage:-

- It performs no actual operation.

- If used accidentally, it can leave required functionality unimplemented.

- It may make a program appear complete when some code is still missing.

```python
for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```

```python
def my_function():
    pass

my_function()
print("Function completed")
```

```python
class Student:
    pass

s1 = Student()
print("Student object created")
```
